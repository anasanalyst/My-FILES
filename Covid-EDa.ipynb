```python
#Making the name much smaller and readable
df= df.rename(columns={"Country/Region":"Countries", "Deaths / 100 Cases": "Death_Cases","Recovered / 100 Cases":"Recovered_Cases",
                      "Deaths / 100 Recovered":"Death_Recovery", "1 week % increase": "1 Week Increase"})
```

```python
# Now we will use the groupby function with mean so that we can see which countires had the top 10 most deaths
mostaffected=df.groupby("Countries")["Deaths"].mean().nlargest(10)
```

```python
mostaffected.reset_index()
```

```python
#using matplotlib and seaborn to visualize the top 10 countires with the most deaths
plt.figure(figsize=(8,6))
plt.style.use("bmh")
sns.set_context("paper", font_scale=1.5)
sns.barplot(x=mostaffected.values, y=mostaffected.index, palette='magma')
plt.title("Top 10 Countries with Most Deaths")
plt.xlabel('Death-Count')
plt.ylabel("Countries")
```

```python
df.head()
```

```python
# Making a boxplt in order to see how the cases increased in 1 week by WHO region
plt.figure(figsize=(10,6))
plt.style.use('bmh')
sns.set_context(font_scale=1)
sns.boxplot(df, x= 'WHO Region', y="1 Week Increase", palette='pastel')
plt.title('Weekly Increase in Covid by WHO region')
plt.xlabel('WHO Region')
plt.ylabel("1 Week Increase(%)")
plt.tight_layout()
```

```python
df.head()
```

```python
#Let see the relationship between total confirmed cases and how many deaths have been occured using the scatter plot.
plt.figure(figsize=(12,6))
plt.style.use('bmh')
sns.scatterplot(df, x='Deaths', y='Recovered', size='Active', hue='WHO Region', alpha=0.7, sizes=(40,400))

plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Lets see the countries which recovered the most
toprecovery=df.groupby('Countries')['Recovered_Cases'].max().nlargest(10)
toprecovery.reset_index()

```

```python
#using the toprecovery in our visualiztion in order to make things easier to see and understand
plt.figure(figsize=(12,8))
plt.style.use('bmh')
sns.barplot(x=toprecovery.values, y=toprecovery.index, palette='Greens_r')
plt.title('10 Most Recovered by Countries')
plt.xlabel('Recovered-Number')
plt.ylabel('Countries')
plt.xlim(0, 100)
plt.tight_layout()
plt.show()
```

```python
#lastly we will make the heatmap and for that we need only the numerical values and their correlation
dfnumerical=df.select_dtypes(include=[np.number])
dfnumerical
```

```python
plt.figure(figsize=(14,7))
plt.style.use('fivethirtyeight')
sns.set_context('paper', font_scale=1.5)
sns.heatmap(dfnumerical.corr(), annot=True, fmt='.2f', cmap='coolwarm', linewidth=2)
plt.title("Correlation HeatMap of Covid-19 Metrics")
plt.tight_layout()
```
