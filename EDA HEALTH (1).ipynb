```python
#Starting the visualization with a heatmap inorder to get the Correlation
#We will make a new dataframe which will be built on numerical columns, since heatmap accepts numeircal columns
#lastly the correlcation will be performed on the numerical columns to get the heatmap
df_num= df.select_dtypes(include=['number'])
```

```python
df_num.corr()
```

```python
#This heatmap is done before without doing any analysis
plt.figure(figsize=(7,5))
plt.style.use('fivethirtyeight')

sns.heatmap(df_num.corr(), cmap='coolwarm', annot=True)
plt.title("Heat-Map")
plt.show()
```

```python
#Lets see the Billing amount based on Admission Type and medical condition
# will use boxplot and violinplot
plt.figure(figsize=(9,5))
plt.style.use('bmh')
sns.set_context('paper', font_scale=1)
sns.barplot(df, x='Admission Type', y='Billing Amount', palette='pastel', estimator=sum)
plt.title("Billing Amount By Admission Type")
```

```python
#Checking the mean of the Billing Amount depending upon the type of admission
billing_amount=df.groupby("Admission Type")['Billing Amount'].mean().reset_index()
billing_amount
```

```python
#Checking the Cost Variation by the medical condition

plt.figure(figsize=(9,5))
plt.style.use('bmh')
plt.cm.RdBu
sns.set_context('paper', font_scale=1)
sns.boxplot(df, x='Medical Condition', y='Billing Amount', palette='deep')
plt.title("Billing Amount By Admission Type")
plt.ylabel("Bill in Dollars--$")
```

```python
cost=df.groupby('Medical Condition')["Billing Amount"].max().reset_index()
cost
```

```python
#Question Patient Demographic: Who is more likely to be Admitted?
#1.Age distribution
#2. Which gender is more likely in emergency 
#3. Age Patteren by Gender
plt.figure(figsize=(8,6))
plt.style.use('bmh')
sns.set_context('paper')
#1
sns.histplot(df['Age'], kde=True)
```

```python
#2
plt.figure(figsize=(8,6))
plt.style.use('bmh')
sns.set_context('paper')

sns.countplot(df, x="Gender", hue="Admission Type", palette="colorblind")
```

```python
plt.figure(figsize=(8,6))
plt.style.use('bmh')
sns.set_context('paper')
3#
sns.violinplot(df, x="Gender", y="Age")
```

```python
#Question 3 Insurance Trends:Which provider covers more high risk patients
#Which Insurers pay more
plt.figure(figsize=(8,6))
plt.style.use("bmh")
sns.set_context("paper", font_scale=1)
sns.boxplot(x="Insurance Provider", y="Billing Amount", data=df)
```

```python
# This plot will show us that which condition is getting more dominant by Insurance
#Business USE : Hospitals can negotiate better rates and insurers adjuting premiums
plt.figure(figsize=(8,6))
plt.style.use('bmh')
sns.set_context('paper', font_scale=1)
sns.countplot(y="Insurance Provider", hue="Medical Condition", data=df, palette='deep')
```

```python
#Question 4 Seasonal Trends: Are there Patterns in hospital admission?
# Business use: Helps hospital prepare for seasonal demand
#  I have already converted the Date of Admission and Discharge date to Date type and now upon this I'll make the countplot

df["Months"]=df["Date of Admission"].dt.month
plt.figure(figsize=(10,5))
plt.style.use('bmh')
sns.set_context("notebook", font_scale=1)
sns.countplot(x="Months", hue='Admission Type', data=df)
plt.title("Monthly Admission Type")
plt.ylabel("Patients Count")
plt.show()
```

```python
#Question 5 Readmission rates: Any condition leading to readmission
sns.countplot(x="Readmitted", hue='Medical Condition',data=df, palette='colorblind')
```
