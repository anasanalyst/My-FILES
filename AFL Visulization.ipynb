```python
#changing the data type of Date and year since they are not accurately in their desired data types
#Changing the dtype of the Date Column in datetime format as it is in the object-dtype.
df['Date']= pd.to_datetime(df['Date'], format="%Y-%m-%d")
```

```python
df.dtypes
```

```python
#changing the dtype of the year as well
df['Year'] = pd.to_datetime(df['Year'], format='%Y')
```

```python
#checking the dtype and now as we are satified we can proceed to the visualization part
df.dtypes
```

```python
#since the attendance got un-necassry ',' and this is causing a problem for us to convert it from object to int
df["Attendance"]=df["Attendance"].str.replace(',', '')
```

```python
df['Attendance']=df["Attendance"].astype(int)
```

```python
df.head(2)
```

```python
# Now we will start the visualization part and visualise the trends using librarires such as Seaborn and Matplotlib
# first we will check the mean/avegrage number of attendance in the stadium by using the groupby and afer all will visualize it
avgdf=df.groupby("Year")['Attendance'].mean().reset_index()
avgdf
```

```python
plt.figure(figsize=(8,6))
plt.style.use('bmh')
sns.set_context('paper')
sns.lineplot(df,x="Year", y="Attendance")
plt.title("Average Attendance Per Year")
```

```python
#Making a KDE Plot to know the Home VS Away Score distribution 
#through this we will know which team was more successfull inscoring goals
plt.figure(figsize=(8,6))
plt.style.use('bmh')
sns.set_context('paper', font_scale=1.5)
sns.kdeplot(df['HomeTeamScore'], label='Home', shade=True)
sns.kdeplot(df['AwayTeamScore'], label="Away", shade=True)
plt.title('Home VS Away')
plt.legend()
plt.show()
```

```python
#3.using temprature to know did it affected the goal scoring of the teams or not 
df['TotalScore']= df['HomeTeamScore'] + df["AwayTeamScore"]
sns.scatterplot(df, x='MaxTemp',y='TotalScore')
plt.title("Temprature vs Score")
#plt.show()
```

```python
df.head(1)
```

```python
#chagning the dtype of attendance and than removing the "," from it
df['Attendance'] = df['Attendance'].replace(',', '', regex=True).astype(float)
df["Attendance"]=df["Attendance"].astype(float)
```

```python
#In order to see the top 10 venue with average attendance we will have to make the mean group by function
avg_attendance=df.groupby("Venue")['Attendance'].mean().head(10)
```

```python
avg_attendance.reset_index()
```

```python

sns.barplot(x=avg_attendance.values, y=avg_attendance.index, palette='magma')
plt.title('Top 10 Venues')
plt.xlabel('Attendance')
plt.show()
```
