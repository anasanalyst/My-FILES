```python
#Starting with Monthly/Weekly activities 
season=df.set_index("InvoiceDate")["Totalsale"].resample("M").sum().reset_index()
season
#Now we will plot a linegrapgh using seaborn to understand the monthly activities
```

```python
#Through this graph we will get to know the actual total Sales done over the past 12 motnhs
#We can check the actual total sales of Days aswell by just replacing the M with D, while making the season calculation
plt.figure(figsize=(9,7))
plt.style.use('bmh')
sns.set_context("paper", font_scale=1.5)
sns.lineplot(data=season, x='InvoiceDate', y='Totalsale', linewidth=1.5)
plt.title("Monthly Sales Over the Year")
plt.show()
```

```python
# lets find the top selling prdducts  with sum of the Totalsale
top_product=df.groupby("Description")['Totalsale'].sum().nlargest(10).reset_index()
top_product
#using seaborn visualization to see these prodcuts in a graph
```

```python
#using seaborn to plot the top 10 most sold products

plt.figure(figsize=(12,6))
plt.style.use("fivethirtyeight")
sns.set_context("talk", font_scale=1)
sns.barplot(top_product, x='Description', y="Totalsale", palette='Set2')
plt.xticks(rotation=70,fontsize=9)
plt.xlabel("Products-Name")
plt.ylabel("Total-Sales")
plt.show()
```

```python
#Finding the Revnue by customer segmentation and they will calssified in high, medium and low category
#Will help us know how many customer fall into each revenue tier
customer_segment=df.groupby("CustomerID")["Totalsale"].sum().reset_index()
customer_segment['Segment']= pd.qcut(customer_segment["Totalsale"], q=3, labels=["LOW", "MEDIUM", "HIGH"])
customer_segment
```

```python
plt.figure(figsize=(10,8))
plt.style.use("bmh")
sns.set_context("paper", font_scale=1)
sns.barplot(data=customer_segment, x='Segment',y="Totalsale", palette='magma')
```

```python
#Total Revenue by customer
revenue_segment=customer_segment.groupby('Segment')["Totalsale"].sum().reset_index()
revenue_segment
sns.barplot(revenue_segment , x="Segment", y="Totalsale", palette="Set2")
plt.ylabel("Total Revenue by Customer Segment")
plt.tight_layout()
```

```python
#Finding the "average order value" over the time (AOV)
df.head(2)
```

```python
df["Month"]=df.loc["InvoiceDate"].dt.to_period("M")
df.head(2)
```

```python
#This shows us the Average Order Value over the 12 Months of the year
aov=df.groupby("Month")["Totalsale"].mean().reset_index()
aov
```

```python
aov["Month"]=aov["Month"].dt.to_timestamp()
```

```python
#This will add a horizntal average line into the grapgh making it more informative
avg=aov["Totalsale"].mean()
```

```python
# Using the Line-plot from sesborn to plot the Average order value 
#Plot
plt.figure(figsize=(12,5))
plt.style.use("bmh")
sns.set_context("paper", font_scale=1)
sns.lineplot(aov, x="Month",y='Totalsale', marker="o", color='red')
#Adding the Title and labels
plt.title("Average Order Value By Month")
plt.xlabel("MONTHS")
plt.ylabel("Average Order By Value")

#Adding the Horiszontal Average-line 
plt.axhline(avg, color="green", linestyle="--", label=f'Overall Avg: ${avg: .2f}')
plt.legend()

#Cleaningup x-axis
plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
#Finding the Sales by the country, Top 10 Countries and skipping United kingdom since it already had a bigger number of sales
filtered_df=df[df["Country"]!='United Kingdom']
Country_sales= (
    filtered_df.groupby("Country")["Totalsale"]
    .sum()
    .nlargest(10)
    .reset_index()
)
```

```python
#Plotting the Graph for the 10 top countires with the most sales using a seaborn barplot
plt.figure(figsize=(12,6))
plt.style.use("bmh")
sns.set_context("paper", font_scale=1)
sns.barplot(Country_sales, x="Country", y="Totalsale", palette='Set2')
plt.title("Top 10 Countries Sales")
plt.xlabel("Countries")
plt.ylabel("Total-Sales")
plt.tight_layout()
plt.show()
```

```python
# Checking the Time of the day which impact on sales, means what time of the day there are high and low sales
df['Hour']=df.loc["InvoiceTime"].dt.hour
```

```python
# calculating the Sum of total sales with the hour's in order to know the sales
hourly_sales=df.groupby('Hour')["Totalsale"].sum().reset_index()
hourly_sales
```

```python
#Plotting a line grapgh in order to understand the sales trend during the time(Hourly).
# Apart from this we will use the Average of Total sales as we used before in the Average Order VAlue
plt.figure(figsize=(12,6))
plt.style.use("fivethirtyeight")
sns.set_context("paper", font_scale=1.5)
sns.lineplot(hourly_sales, x="Hour", y='Totalsale',marker='o', color='royalblue', linewidth=2)
#Adding the title and the labels
plt.title("Hourly Sales")
plt.ylabel('Total Sales With Time')

# Add formatted data labels with 1 decimal point
for i in range(len(hourly_sales)):
    value = hourly_sales['Totalsale'][i]
    hour = hourly_sales['Hour'][i]
    plt.text(
        hour,
        value + max(hourly_sales['Totalsale']) * 0.02,  # raise text a bit higher
        f"{value:,.1f}",  # 1 decimal point
        ha='center',
        fontsize=9,
        bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="gray", lw=0.5)
    )

plt.tight_layout()
plt.show()
```
